```python
import pandas as pd
import plotly.express as px

hourly_df = pd.read_csv('../data/processed/hourly_energy.csv')
hourly_df['timestamp'] = pd.to_datetime(hourly_df['timestamp'])
hourly_df.head()
```

```python
fig = px.line(
    hourly_df, 
    x='timestamp', 
    y='Global_active_power',
    title='Energy Consumption Over Time'
)
fig.show()
```

```python
hourly_avg = hourly_df.groupby('hour')['Global_active_power'].mean().reset_index()

fig = px.bar(
    hourly_avg, 
    x='hour', 
    y='Global_active_power',
    title='Average Energy Consumption by Hour of Day'
)
fig.show()
```

```python
hourly_df['day_type'] = hourly_df['is_weekend'].map({True: 'Weekend', False: 'Weekday'})

day_avg = hourly_df.groupby('day_type')['Global_active_power'].mean().reset_index()

fig = px.bar(
    day_avg, 
    x='day_type', 
    y='Global_active_power',
    title='Weekday vs Weekend Consumption'
)
fig.show()
```

```python
monthly_avg = hourly_df.groupby('month')['Global_active_power'].mean().reset_index()

fig = px.bar(
    monthly_avg, 
    x='month', 
    y='Global_active_power',
    title='Average Energy Consumption by Month'
)
fig.show()
```

```python
categories = pd.DataFrame({
    'Category': ['Kitchen', 'Laundry', 'Water Heater/AC'],
    'Total_Consumption': [
        hourly_df['Sub_metering_1'].sum(),
        hourly_df['Sub_metering_2'].sum(),
        hourly_df['Sub_metering_3'].sum()
    ]
})

fig = px.pie(
    categories, 
    names='Category', 
    values='Total_Consumption',
    title='Energy Consumption by Category'
)
fig.show()
```
